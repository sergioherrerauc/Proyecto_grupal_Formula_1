```python
import pandas as pd

# 1. Indicamos el nombre del archivo
file_name = 'Base_de_datos_Mariana_2_limpia.csv'

# 2. Configurar la herramienta Pandas para forzar la visualización completa
pd.set_option('display.max_rows', None)     # Muestra todas las filas (años)
pd.set_option('display.max_columns', None)  # Muestra todas las columnas (posiciones)
pd.set_option('display.width', None)        # Adapta el ancho sin cortar el texto

# 3. Leer el archivo CSV
df_mariana = pd.read_csv(file_name)

# 4. Imprimir un resumen de las categorías (estructura técnica)
print("--- Estructura y Categorías de la Base de Datos ---")
df_mariana.info()
print("\n" + "="*80 + "\n")

# 5. Visualizar todo el contenido de la tabla en pantalla
print("--- Visualización Completa de Datos ---")
display(df_mariana)
```
