```python
# Proyecto F1 2015-2019 Sergio Herrera

import altair as alt
import pandas as pd

# 1. Base de datos a usar
df = pd.read_csv("database_limpia_F1_2015_2019.csv")

# 2. Gráfico indivual más los años individuales
df["Año_Filtro"] = df["Año"].astype(str)
df_global = df.copy()
df_global["Año_Filtro"] = "Global (2015-2019)"
df_combinado = pd.concat([df, df_global], ignore_index=True)

# 3. Seleccion del año temporada
select_year = alt.binding_select(
    options=[
        "Global (2015-2019)",
        "2015",
        "2016",
        "2017",
        "2018",
        "2019",
    ],
    name="Temporada: ",
)
year_selection = alt.selection_single(fields=["Año_Filtro"], bind=select_year)

chart = (
    alt.Chart(df_combinado)
    .mark_bar(cornerRadiusTopRight=4, cornerRadiusBottomRight=4)
    .encode(
        x=alt.X(
            "Presupuesto_Millones_USD:Q",
            title="Presupuesto Anual (Millones de USD)",
            axis=alt.Axis(
                grid=True,
                titleFontSize=14,
                labelFontSize=12,
                titleColor="#2c3e50",
                labelColor="#34495e",
            ),
        ),
        y=alt.Y(
            "Equipo:N",
            sort="-x",
            title="Escudería",
            axis=alt.Axis(
                titleFontSize=14,
                labelFontSize=12,
                titleColor="#2c3e50",
                labelColor="#34495e",
            ),
        ),
        color=alt.Color(
            "Categoria_Equipo:N",
            title="Categoría",
            scale=alt.Scale(
                domain=["Top 3", "Zona Media", "Zona Baja"],
                range=["#e74c3c", "#3498db", "#2ecc71"],
            ),
            legend=alt.Legend(
                titleFontSize=13, labelFontSize=12, symbolSize=130, padding=10
            ),
        ),
        tooltip=[
            alt.Tooltip("Equipo:N", title="Escudería"),
            alt.Tooltip("Año_Filtro:N", title="Temporada"),
            alt.Tooltip(
                "Presupuesto_Millones_USD:Q", title="Presupuesto (mUSD)"
            ),
            alt.Tooltip("Puntos:Q", title="Puntos FIA"),
            alt.Tooltip(
                "Costo_Por_Punto_Millones_USD:Q", title="Costo por Punto (mUSD)"
            ),
            alt.Tooltip("Categoria_Equipo:N", title="Categoría"),
        ],
    )
    .add_selection(year_selection)
    .transform_filter(year_selection)
    .properties(
        title={
          #4. Narrativa del gráfico inicial
               "text": [
                '"Cuando un punto cuesta una fortuna": La inversión de la F1 en la era del gasto libre'
            ],
            "subtitle": [
                "Desglose global 2015 a 2019. Selecciona en el menú de abajo 'Global' o una temporada específica para comparar el impacto financiero de las escuderías."
            ],
        },
        width=750,
        height=420,
    )
    .configure_title(
        fontSize=18,
        font="Helvetica, Arial, sans-serif",
        anchor="start",
        color="#111827",
        subtitleFontSize=13,
        subtitleColor="#4b5563",
        subtitlePadding=8,
    )
    .configure_axis(
        labelFont="Helvetica, Arial, sans-serif",
        titleFont="Helvetica, Arial, sans-serif",
        titleFontWeight="bold",
    )
    .configure_legend(
        titleFont="Helvetica, Arial, sans-serif",
        labelFont="Helvetica, Arial, sans-serif",
        titleFontWeight="bold",
    )
    .configure_view(stroke=None)
)
#5. Guardar vista en HTML
chart.save("vis_01.html")
```
